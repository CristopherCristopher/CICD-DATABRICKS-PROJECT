```sql
-- ============================================================
--  VALIDAR TABLAS GOLDEN
-- ============================================================

SHOW TABLES IN catalog_au.golden;

SELECT *
FROM catalog_au.golden.transport_comparison
LIMIT 10;
```

```sql
-- ============================================================
--  CREAR DELTA SHARE
-- ============================================================

CREATE SHARE mobility_share
COMMENT 'Datos Golden del proyecto NYC Multimodal Urban Mobility';
```

```sql
-- ============================================================
--  VALIDAR SHARE
-- ============================================================

SHOW SHARES;
```

```sql
-- ============================================================
-- COMPARTIR MOBILITY HOURLY
-- ============================================================

ALTER SHARE mobility_share
ADD TABLE catalog_au.golden.mobility_hourly
WITH HISTORY;
```

```sql
-- ============================================================
-- COMPARTIR MOBILITY WEATHER
-- ============================================================

ALTER SHARE mobility_share
ADD TABLE catalog_au.golden.mobility_weather
WITH HISTORY;
```

```sql
-- ============================================================
-- COMPARTIR MOBILITY ZONE
-- ============================================================

ALTER SHARE mobility_share
ADD TABLE catalog_au.golden.mobility_zone
WITH HISTORY;
```

```sql
-- ============================================================
-- COMPARTIR TRANSPORT COMPARISON
-- ============================================================

ALTER SHARE mobility_share
ADD TABLE catalog_au.golden.transport_comparison
WITH HISTORY;

```

```sql
-- ============================================================
-- VALIDAR CONTENIDO DEL SHARE
-- ============================================================

SHOW ALL IN SHARE mobility_share;
```

```sql
-- ============================================================
-- CREAR RECIPIENT PARA POWER BI
-- ============================================================

CREATE RECIPIENT powerbi_recipient
COMMENT 'Recipient externo para consumo de Golden desde Power BI';
```

```sql
-- ============================================================
-- VALIDAR RECIPIENT
-- ============================================================

SHOW RECIPIENTS;
```

```sql
-- ============================================================
-- OTORGAR SHARE AL RECIPIENT
-- ============================================================

GRANT SELECT
ON SHARE mobility_share
TO RECIPIENT powerbi_recipient;
```

```sql
-- OBTENER DATOS DEL RECIPIENT
-- ============================================================

DESCRIBE RECIPIENT powerbi_recipient;
```

```sql
-- ============================================================
-- VALIDAR ACCESO DEL RECIPIENT
-- ============================================================

SHOW GRANTS TO RECIPIENT powerbi_recipient;
```

```sql
-- ============================================================
-- VALIDAR RECIPIENTS CON ACCESO AL SHARE
-- ============================================================

SHOW GRANTS ON SHARE mobility_share;
```

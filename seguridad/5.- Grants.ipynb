```sql
-- ============================================================
-- DAs - DATA ANALYSTS
-- ACCESO SOLO LECTURA A GOLDEN
-- ============================================================

GRANT USE CATALOG
ON CATALOG catalog_au
TO `DAs`;


GRANT USE SCHEMA
ON SCHEMA catalog_au.golden
TO `DAs`;


GRANT SELECT
ON SCHEMA catalog_au.golden
TO `DAs`;
```

```sql
-- ============================================================
-- DEs - DATA ENGINEERS
-- ============================================================

GRANT USE CATALOG
ON CATALOG catalog_au
TO `DEs`;

-- por schemas
GRANT USE SCHEMA
ON SCHEMA catalog_au.bronze
TO `DEs`;


GRANT USE SCHEMA
ON SCHEMA catalog_au.silver
TO `DEs`;


GRANT USE SCHEMA
ON SCHEMA catalog_au.golden
TO `DEs`;
```

```sql
-- ============================================================
-- DEs - PERMISOS SOBRE TABLAS
-- ============================================================

GRANT SELECT, MODIFY
ON SCHEMA catalog_au.bronze
TO `DEs`;


GRANT SELECT, MODIFY
ON SCHEMA catalog_au.silver
TO `DEs`;


GRANT SELECT, MODIFY
ON SCHEMA catalog_au.golden
TO `DEs`;
```

```sql
-- ============================================================
-- DEs - LECTURA DE RAW
-- ============================================================

GRANT READ FILES
ON EXTERNAL LOCATION `exlt-raw`
TO `DEs`;

```

```sql
-- ============================================================
-- PAs - PLATFORM ADMINISTRATORS
-- ============================================================

GRANT ALL PRIVILEGES
ON CATALOG catalog_au
TO `PAs`;

GRANT MANAGE
ON CATALOG catalog_au
TO `PAs`;
```

```sql
-- ============================================================
-- PAs - EXTERNAL LOCATIONS
-- ============================================================

GRANT ALL PRIVILEGES
ON EXTERNAL LOCATION `exlt-raw`
TO `PAs`;

GRANT MANAGE
ON EXTERNAL LOCATION `exlt-raw`
TO `PAs`;


GRANT ALL PRIVILEGES
ON EXTERNAL LOCATION `exlt-bronze`
TO `PAs`;

GRANT MANAGE
ON EXTERNAL LOCATION `exlt-bronze`
TO `PAs`;


GRANT ALL PRIVILEGES
ON EXTERNAL LOCATION `exlt-silver`
TO `PAs`;

GRANT MANAGE
ON EXTERNAL LOCATION `exlt-silver`
TO `PAs`;


GRANT ALL PRIVILEGES
ON EXTERNAL LOCATION `exlt-golden`
TO `PAs`;

GRANT MANAGE
ON EXTERNAL LOCATION `exlt-golden`
TO `PAs`;
```
